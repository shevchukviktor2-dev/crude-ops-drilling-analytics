```python
import pytest
from pyspark.sql import SparkSession

def test_gold_table_not_empty():
    spark = SparkSession.builder.getOrCreate()
    count = spark.table("crude_ops.gold.drilling_stats_summary").count()
    assert count > 0, "Gold table should have data!"

def test_gamma_ray_range():
    spark = SparkSession.builder.getOrCreate()
    # check that gamma ray is in range [0, 100] (it is possible to get negative values)
    invalid_count = spark.table("crude_ops.gold.drilling_stats_summary").filter("avg_gamma_ray < 0").count()
    assert invalid_count == 0, "Found negative Gamma Ray values!"
```
