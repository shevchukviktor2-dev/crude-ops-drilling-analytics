```python
folder_path = "/Volumes/crude_ops/bronze/oil_landing_zone/"
target_table = "crude_ops.bronze.sap_operations"

try:
    # 1. Read files
    sap_raw_df = spark.read.format("csv") \
        .option("header", "true") \
        .option("inferSchema", "true") \
        .load(f"{folder_path}/*.csv")

    # 3. Save to Bronze layer
    # Overwrite the table with the full data set from the folder
    sap_raw_df.write.format("delta") \
        .mode("overwrite") \
        .option("overwriteSchema", "false") \
        .saveAsTable(target_table)

    print(f"SUCCESS: Schema updated and data loaded into {target_table}")
    display(spark.table(target_table).limit(5))

except Exception as e:
    print(f"Error during ingestion: {str(e)}")
```
